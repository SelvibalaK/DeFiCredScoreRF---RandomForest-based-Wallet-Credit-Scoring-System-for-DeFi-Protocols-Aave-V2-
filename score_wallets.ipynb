```python
import pandas as pd
```

```python
df = pd.read_csv(r"D:\User1\deficreditscore\wallet_features.csv")
```

```python
df["score"] = 500
```

```python
df.loc[df["repay_ratio"] >= 1.0, "score"] += 200  # Good repayers
df.loc[df["repay_ratio"] < 0.5, "score"] -= 150   # Bad repayers
```

```python
df.loc[df["total_liquidations"] > 0, "score"] -= 100 * df["total_liquidations"]
```

```python
df.loc[df["days_active"] > 90, "score"] += 100
df.loc[df["days_active"] < 7, "score"] -= 50
```

```python
df.loc[df["unique_actions"] <= 2, "score"] -= 50
```

```python
df.loc[(df["total_borrow"] > 1e6) & (df["repay_ratio"] >= 1.0), "score"] += 100
```

```python
df["score"] = df["score"].clip(0, 1000)
```

```python
df[["wallet", "score"]].to_csv("wallet_scores.csv", index=False)
print("✅ Wallet scoring complete. Output saved to wallet_scores.csv")
```
