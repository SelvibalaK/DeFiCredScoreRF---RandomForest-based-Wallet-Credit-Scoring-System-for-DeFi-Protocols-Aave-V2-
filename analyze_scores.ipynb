```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
features = pd.read_csv(r"D:\User1\deficreditscore\wallet_features.csv")
scores = pd.read_csv(r"D:\User1\deficreditscore\wallet_scores.csv")
df = features.merge(scores, on="wallet")
```

```python
plt.figure(figsize=(10, 6))
bins = list(range(0, 1100, 100))
sns.histplot(df["score"], bins=bins, kde=False, color='skyblue', edgecolor='black')
plt.title("Wallet Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Number of Wallets")
plt.grid(True)
plt.xticks(bins)
plt.savefig("score_distribution.png")
plt.show()
```

```python
df["score_band"] = pd.cut(df["score"], bins=bins, right=False)
summary = df.groupby("score_band")[["total_borrow", "total_repay", "repay_ratio", "total_liquidations", "days_active"]].mean().round(2)
summary.to_csv("score_band_summary.csv")
print("✅ Analysis complete. Score distribution image and summary CSV saved.")
```
